```python
import pandas as pd
import numpy as np

# ১. একটি স্যাম্পল ডেটাফ্রেম তৈরি (আপনার ডেটার মতো)
data = {
    'Name': ['John', 'Akane', 'Doe'],
    'Score_Text': ['85', 'John', '90'], # এখানে মাঝখানে 'John' ভুল করে ঢুকেছে
    'Attendance': [10, 20, 30]
}
df = pd.DataFrame(data)

# ২. কলাম যোগ করার আগে ডেটা টাইপ ঠিক করা
# errors='coerce' দিলে 'John' এর মতো নামগুলো NaN (Not a Number) হয়ে যাবে
df['Score_Numeric'] = pd.to_numeric(df['Score_Text'], errors='coerce')

# ৩. নতুন কলাম যোগ করা (গাণিতিক হিসাবসহ)
# ধরা যাক, আমরা Score এবং Attendance যোগ করে একটি Total কলাম বানাচ্ছি
df['Total'] = df['Score_Numeric'] + df['Attendance']

# ৪. ফালতু বা NaN ভ্যালুগুলো ০ দিয়ে ভর্তি করা (ঐচ্ছিক)
df['Total'] = df['Total'].fillna(0)

print("--- আপডেট হওয়া ডেটাফ্রেম ---")
print(df)
```
