```python
import pandas as pd 
import numpy as np

# ১. ডেটা কপি করা
data_set2 = data_set1.copy()
data_se = data_set1[2:4].copy()

# ২. লুপ চালিয়ে নির্দিষ্ট কিছু ঘরে NaN বসানো
for j in range(data_se.shape[1]): # প্রতিটি কলামের জন্য
    # ৩টি র‍্যান্ডম সারির ইনডেক্স বেছে নেওয়া
    rows_to_nan = np.random.choice(data_se.shape[0], 3, replace=True)
    
    # .iloc ব্যবহার করে ঐ নির্দিষ্ট সারি ও কলামে NaN বসানো
    data_se[rows_to_nan, j] = np.nan

# ৩. মূল ডেটাসেটের নির্দিষ্ট অংশে এই পরিবর্তিত ডেটা বসানো
data_set2[2:4] = data_se

# ৪. ফলাফল প্রিন্ট করা
print(data_set2[3:5]) # পরিবর্তন দেখার জন্য ৬০-৭৫ সারি প্রিন্ট
```
