```python
plt.plot(batman['index'],batman['V Kohli'],color='red',linestyle='dashdot',marker='+',markersize=10,label='vorat')
plt.plot(batman['index'],batman['RG Sharma'],color='green',linestyle='dashdot',marker='+',markersize=10,label='Sharma')
plt.title('Rg sharma vs virat kohil career')
plt.xlabel('seasson')
plt.ylabel('Runs Scored')
plt.legend('upper right')
```

```python
df=pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2,color="red")
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color="green")
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2,color="black")
plt.xlabel('batman')
plt.ylabel('year')
plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2,color="green")
plt.bar(np.arange(df.shape[0]),df["2016"],width=0.2)
plt.bar(np.arange(df.shape[0])+0.2,df["2017"],width=0.2)
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2,color="red")
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color="green")
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2,color="black")
plt.xlabel('batman')
plt.ylabel('year')
plt.xticks(np.arange(df.shape[0]),df['batsman'])
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2,color="red")
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color="green")
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2,color="black")
plt.xlabel('batman')
plt.ylabel('year')
plt.xticks(rotation='vertical')
```

```python
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016']+df['2017']),label='2015')
plt.legend()
plt.show()
```
