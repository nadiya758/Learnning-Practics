# Sub plot

Using fig, ax = plt.subplots() is the industry-standard way to use Matplotlib. It triggers the Object-Oriented (OO) interface, which is much more powerful and cleaner than the basic plt.plot() (state-machine) approach.

```python
batters.sample(5)
```

```python
plt.figure(figsize=(15,6))
plt.scatter(batters['avg'],batters['strike_rate'])
plt.title('Somrthing')
plt.xlabel('Avg')
plt.ylabel('strika_rate')
```

```python
fig,ax = plt.subplots(figsize=(15,6))
ax.scatter(batters['avg'],batters['strike_rate'],color='red',marker='+')
ax.set_title('Something')
ax.set_xlabel('avg')
ax.set_ylabel('stirika')
```

```python

fig, ax = plt.subplots(nrows=2,ncols=1,figsize=(10,8),sharex=True)
ax[0].scatter(batters['avg'],batters['strike_rate'],color='red',marker='D')
ax[1].scatter(batters['avg'],batters['runs'])
ax[0].set_title('strike and Runs')
ax[0].set_ylabel('Strike_Rate')
ax[1].set_title('strike and Runs')
ax[1].set_xlabel('Avg')
ax[1].set_ylabel('Runs')
```

```python

fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10))
ax[0,0].scatter(batters['avg'],batters['strike_rate'],color='red',marker='D')
ax[1,0].scatter(batters['avg'],batters['runs'])
ax[0,1].hist(batters['avg'])
ax[1,1].hist(batters['runs'])
ax[0,0].set_title('strike and Runs')
ax[0,0].set_ylabel('Strike_Rate')
ax[1,0].set_title('strike and Runs')
ax[1,0].set_xlabel('Avg')
ax[1,0].set_ylabel('Runs')
for i in range(batters.shape[0]):
    ax[0,0].text(batters['avg'].values[i],batters['strike_rate'].values[i],batters['runs'].values[i])
plt.tight_layout()
plt.show()

```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(nrows=2, ncols=2, figsize=(10,10))

# ১. প্রথম স্ক্যাটার প্লট (ax[0,0])
ax[0,0].scatter(batters['avg'], batters['strike_rate'], color='red', marker='D')
ax[0,0].set_title('Avg vs Strike Rate')
ax[0,0].set_ylabel('Strike_Rate')
for i in range(batters.shape[0]):
    ax[0,0].text(batters['avg'].values[i], batters['strike_rate'].values[i], str(batters['runs'].values[i]))

# ২. দ্বিতীয় স্ক্যাটার প্লট (ax[1,0])
ax[1,0].scatter(batters['avg'], batters['runs'])
ax[1,0].set_title('Avg vs Runs')
ax[1,0].set_xlabel('Avg')
ax[1,0].set_ylabel('Runs')

# ৩. প্রথম হিস্টোগ্রাম (ax[0,1]) - এখানে টেক্সট অ্যাড করা হয়েছে
n1, bins1, patches1 = ax[0,1].hist(batters['avg'], color='skyblue', edgecolor='black')
ax[0,1].set_title('Avg Histogram')
for i in range(len(n1)):
    if n1[i] > 0:
        ax[0,1].text(bins1[i] + (bins1[i+1]-bins1[i])/2, n1[i] + 0.1, int(n1[i]), ha='center', va='bottom')

# ৪. দ্বিতীয় হিস্টোগ্রাম (ax[1,1]) - এখানেও টেক্সট অ্যাড করা হয়েছে
n2, bins2, patches2 = ax[1,1].hist(batters['runs'], color='lightgreen', edgecolor='black')
ax[1,1].set_title('Runs Histogram')
for i in range(len(n2)):
    if n2[i] > 0:
        ax[1,1].text(bins2[i] + (bins2[i+1]-bins2[i])/2, n2[i] + 0.1, int(n2[i]), ha='center', va='bottom')

plt.tight_layout()
plt.show()
```
